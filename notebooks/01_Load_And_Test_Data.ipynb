```python
import pandas as pd, sqlite3
con = sqlite3.connect("../grocery.db")
for name in ["train", "stores", "transactions", "holidays_events", "oil"]:
    pd.read_csv(f"../data/{name}.csv").to_sql(name, con, if_exists="replace", index=False)
print("done")
```

```python
pd.read_sql("SELECT name FROM sqlite_master WHERE type='table'", con)
```

```python
pd.read_sql("SELECT MIN(date), MAX(date), COUNT(*) FROM train", con)
```

```python
pd.read_sql("""
SELECT strftime('%Y', date) AS yr,
       ROUND(AVG(onpromotion > 0), 3) AS share_with_promo
FROM train GROUP BY yr
""", con)
```

```python
pd.read_sql("""
SELECT COUNT(DISTINCT date) AS days_present,
       CAST(julianday(MAX(date)) - julianday(MIN(date)) + 1 AS INT) AS days_in_range
FROM train
""", con)
```

```python
pd.read_sql("SELECT MIN(date) FROM train WHERE onpromotion > 0", con)
```
